3.Numerical Operators

```r
x <-10
y <- 3

add <- x + y
sub <- x - y
mul <- x * y
div <- x / y
mod <- x %% y
exp <- x ^ y


print(add)
print(sub)
print(mul)
print(div)
print(mod)
print(exp)
```

11.DataFrame

```r
students <- data.frame(
  name = c("Aaron","Jim","Cueto"),
  course = c("BSCS","BSIT","BSA"),
  grade = c(75,77,90)
)
students
```
