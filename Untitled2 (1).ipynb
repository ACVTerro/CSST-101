8.Creating a Data Frame

```r
students <- data.frame(
  Student_ID = c("S001","S002","S003","S004","S005"),
  Attendance = c(95,72,88,79,68),
  Grade = c(90,70,85,77,65)
)
students
```

9.Addin a classification Column

```r
students$Performance <- ifelse(
  students$Grade >= 80,
  "Good",
  ifelse(
    students$Grade >= 75,
    "passed",
    "failed"
  )
)
students
```

10.Creating a Simple Rule-Based System

```r
students$risk <- ifelse(
  students$Attendance < 75 & students$Grade < 75,
  "High risk",
  "low risk"
)

students
```

11.Add More Rules

```r
student_risk <- function(attendance,grade){
if (attendance < 75 && grade < 75){
  print("High risk")
} else if (attendance < 85 || grade < 80){
  print("Moderate Risk")
} else if (attendance >= 85 && grade >= 80){
  print("low risk")
}
}

students$risk <- mapply(
  student_risk,
  students$Attendance,
  students$Grade
                        )
students
```

12.Count the Results

13.create a visualization

```r
risk_counts <- table(students$risk)

barplot(
  risk_counts,
  main = "Student Risk Classification",
  xlab = "Risk level",
  ylab = "Number of Students"
)

print("1.that there are both more low and high risk students than moderate risk")
print("2.both high and low risk")
print("3.encourage for the help of the low risk students to help the high risk")
```

14.Add a recommendation

```r
students$recommendation <- ifelse(
  students$risk == "High risk",
  "Immediate Academic Intervention",
  ifelse(
    students$risk == "Moderate risk","Academic Monitoring","Continue Regular Monitoring"
))
students
```
